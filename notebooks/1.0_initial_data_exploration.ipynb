# Initial Exploratory Data Analysis (EDA)

This notebook provides an EDA on the Lung Cancer dataset, including importing data to MongoDB and exploring the dataset structure.

## Dataset Overview
- **Name**: Lung Cancer Dataset
- **Source**: [Kaggle](https://www.kaggle.com/datasets/akashnath29/lung-cancer-dataset?select=dataset.json)
- **Description**: This dataset is focused on detecting lung cancer and includes various features associated with common symptoms and risk factors.

## Importing Necessary Libraries and Configuration

Here, we load `pymongo` for MongoDB connections and `json` for handling JSON data.

```python
import json
from pymongo import MongoClient
import sys
sys.path.append('..') 
import scripts.utils  

client = scripts.utils.connect_to_mongo()
```

## Importing the Data to MongoDB

Load JSON data and insert it into MongoDB for analysis.

```python
# Load dataset from the path defined in utils
dataset_path = scripts.utils.DATA_PATH  # Get the relative path from utils.py
data = scripts.utils.load_json_data(dataset_path)

# Insert data into MongoDB
db = client['Lungcancer']
collection = db['Patients']
result = collection.insert_many(data)

print(f'Inserted {len(result.inserted_ids)} documents.')
```

## Exploring the Data

Here we explore the dataset by checking the number of collections and documents, analyzing the schema, and viewing a sample document.

```python
# Count the number of collections and documents
num_collections = len(db.list_collection_names())
print(f"Number of collections: {num_collections}")

num_documents = collection.count_documents({})
print(f"Number of documents in 'Patients' collection: {num_documents}")
```

```python
# Display the schema by analyzing the types of fields in a sample document
schema = {}
sample_docs = collection.find().limit(5)  
for doc in sample_docs:
    for key, value in doc.items():
        schema[key] = type(value).__name__

print("\nSchema of the dataset:")
for key, value in schema.items():
    print(f"{key}: {value}")
```

```python
# Fetch and print a sample document from the dataset
sample_document = collection.find_one()
print("\nSample document:")
print(sample_document)
```

## Key Fields in the Dataset
- **LUNG_CANCER**: Diagnosis of lung cancer; critical for predictive modeling.
- **SMOKING**: Smoking status; major risk factor.
- **AGE**: Patient's age; higher risk with age.
- **GENDER**: Patient's gender; may impact risk.
- **CHRONIC_DISEASE**: Presence of chronic diseases; impacts lung health.
- **CHEST_PAIN**: Presence of chest pain; key symptom.
- **SHORTNESS_OF_BREATH**: Symptom indicating lung issues.
